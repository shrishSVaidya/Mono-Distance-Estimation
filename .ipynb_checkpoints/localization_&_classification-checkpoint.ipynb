```python
import torch
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

# Model loading

```python
model = torch.hub.load('ultralytics/yolov5', 'yolov5s')
```

```python
model
```

```python
img = 'https://www.constructionweekonline.in/cloud/2022/03/07/office_space_kj3wup.jpg'
```

```python
results = model(img)
```

```python
results.print()
```

```python
%matplotlib inline
plt.figure(figsize=(20, 20))
plt.imshow(np.squeeze(results.render()))
plt.show()
```

```python
results.xyxy    # xmin, ymin, xmax, ymax, confidence, class
```

# Real-time

```python
cap = cv2.VideoCapture(0)
```

```python
while True:
    rt, frame = cap.read()
   
    print(frame.shape)
    results = model(frame)
    
    cv2.imshow('Frame', np.squeeze(results.render()))
#     cv2.imshow('Frame', frame)
   
    
    if cv2.waitKey(1)==13:
        break
        
cap.release()
cv2.destroyAllWindows()
```
