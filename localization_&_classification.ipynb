# Real-time

```python
cap = cv2.VideoCapture(0)
```

```python
while True:
    rt, frame = cap.read()
   
    print(frame.shape)
    results = model(frame)
    
    cv2.imshow('Frame', np.squeeze(results.render()))
#     cv2.imshow('Frame', frame)
   
    
    if cv2.waitKey(1)==13:
        break
        
cap.release()
cv2.destroyAllWindows()
```

```python
model = torch.hub.load('ultralytics/yolov5', 'custom', path = 'yolov5/runs/train/exp10/weights/last.pt', force_reload=True)
```

```python
import os
img = os.path.join('data', 'images', '105.jpg')
results = model(img)
results.print()
```

```python
%matplotlib inline
plt.figure(figsize=(20, 20))
plt.imshow(np.squeeze(results.render()))
plt.show()
```

```python

results.xyxy[0].tolist()[0] # xmin, ymin, xmax, ymax, confidence, class
```

```python
import math
def calc(obj):
    xmin, ymin, xmax, ymax, confidence, clas = obj.xyxy[0].tolist()[0]
    width = (xmax-xmin)
    length = (ymax-ymin)
    area = width*length
    diagonal = math.hypot(width, length)
    print(f'length = {length}, width = {width}, area = {area}, diagonal = {diagonal}')
```

```python
calc(results)
```
